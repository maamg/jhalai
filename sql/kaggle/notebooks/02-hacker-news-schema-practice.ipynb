```python
from google.cloud import bigquery
```

```python
# Creating a Client object
client= bigquery.Client()

```

```python
# Create a reference to a dataset

dataset_ref = client.dataset('hacker_news', project= 'bigquery-public-data')

# get a dataset into a variable
dataset = client.get_dataset(dataset_ref)

```

```python
# List all the tables in the "hacker_news" dataset
table_list= list(client.list_tables(dataset))
for each_table in table_list:
    print(each_table.table_id)
```

**এখন এই full টেবিল টা নিতে হবে ****

```python
# প্রথমে টেবিলের রেফারেন্স তৈরি করি
table_ref= dataset_ref.table('full')
tableFull = client.get_table(table_ref)
```

```python
tableFull.schema 
```

```python
columnNames= [column.name for column in tableFull.schema]
print(columnNames)
```

```python
for eachColumn in tableFull.schema: 
    print(eachColumn.name, '| ', eachColumn.field_type, '| ')
```

```python
client.list_rows(tableFull, max_results= 7).to_dataframe()
```

```python
client.list_rows(tableFull, selected_fields=tableFull.schema[:3], max_results=5).to_dataframe()
```
